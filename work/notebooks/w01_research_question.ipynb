## 3. Quick look at the data (2-3 real numbers)

###**Key Data Evidence for the AI Referral Opportunity Track**
1. **AI Signal Sparsity Rate:** Only ~1% to 2% of all indexed pages currently receive non-zero AI referral sessions (*ai_sessions_90d > 0*), confirming that AI citation is a highly specialized, sparse signal.

2. **The AI Opportunity Gap:** Over 95% of high-impression organic pages (*impressions_90d >= 500*) currently capture zero AI referral sessions, representing a massive untapped optimization window for content teams.

3. **Actionable High-Potential Pool:** There are thousands of deep, high-engagement pages (word_count >= 800, above-median scroll rate) with solid search demand that have zero AI traffic—forming our prime target queue for Generative Engine Optimization (GEO).

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd
import numpy as np

# 1. Load the anonymized starter dataset
data_path = r"C:\Users\prasa\Projects\Flyrank-ML\machinelearning01-flyrank\data\raw\content_refresh_anonymized.csv"
df = pd.read_csv(data_path)

# Safe cleaning: fill null AI session counts with 0
df["ai_sessions_90d"] = df["ai_sessions_90d"].fillna(0)

#AI Traffic Sparsity Rate
total_pages = len(df)
pages_with_ai = (df["ai_sessions_90d"] > 0).sum()
pct_pages_with_ai = (pages_with_ai / total_pages) * 100

#The Opportunity Gap in High-Visibility Content
high_organic_df = df[df["impressions_90d"] >= 500]
high_organic_total = len(high_organic_df)
high_organic_zero_ai = (high_organic_df["ai_sessions_90d"] == 0).sum()
pct_zero_ai = (high_organic_zero_ai / high_organic_total) * 100 if high_organic_total > 0 else 0

#High-Potential Candidate Pool
median_scroll = df["scroll_rate"].median()
high_potential_df = df[
    (df["impressions_90d"] >= 100) &
    (df["ai_sessions_90d"] == 0) &
    (df["word_count"] >= 800) &
    (df["scroll_rate"] >= median_scroll)
]
num_candidates = len(high_potential_df)

#Output report
print("=" * 76)
print("       AI REFERRAL LANE: DATA AUDIT PROOF POINTS       ")
print("=" * 76)
print(f"Total Dataset Pages Evaluated : {total_pages:,}")
print(f"1. Pages with >0 AI Sessions   : {pages_with_ai:,} ({pct_pages_with_ai:.2f}% sparsity rate)")
print(f"2. High-Organic Pages Zero AI : {high_organic_zero_ai:,} / {high_organic_total:,} ({pct_zero_ai:.1f}% opportunity gap)")
print(f"3. High-Potential GEO Pool    : {num_candidates:,} eligible pages ready for optimization")
print("=" * 76)
```
